```python
# Dot Product
def vector_dot_product(v1, v2):
    if len(v1) != len(v2):
        return "Vectors must have the same length"

    return sum(a * b for a, b in zip(v1, v2))


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = vector_dot_product(v1, v2)
print(f"Dot Product of {v1} & {v2} is {result}")
```

```python
# Magnitude / Norm / Length
def vector_norm(v):
    if not v:
        return "Vector is empty"
    if not all(isinstance(x, (int, float)) for x in v):
        return "Vector must contain only numbers"
    return (sum(x * x for x in v)) ** 0.5


v = [1, 5, 9]

result = vector_norm(v)
print(f"Norm of {v} is {round(result, 2)}")
```

```python
# Vector Addition
def vector_add(v1, v2):
    if len(v1) != len(v2):
        return "Vectors must have the same length"

    return [a + b for a, b in zip(v1, v2)]


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = vector_add(v1, v2)
print(f"Sum of {v1} & {v2} is {result}")
```

```python
# Vector Subtraction
def vector_subtract(v1, v2):
    if len(v1) != len(v2):
        return "Vectors must have the same length"

    return [a - b for a, b in zip(v1, v2)]


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = vector_subtract(v1, v2)
print(f"Subtraction of {v1} & {v2} is {result}")
```

```python
# Scalar Multiplication
def scalar_multiply(v, k):
    return [k * x for x in v]


k = 5
v = [1, 5, 9]

result = scalar_multiply(v, k)
print(f"Scalar multiplication of {v} & {k} is {result}")
```

```python
# Distance Between Two Vectors
def distance_between_vectors(v1, v2):
    if len(v1) != len(v2):
        return "Vectors must have the same length"
    diff = [a - b for a, b in zip(v1, v2)]
    return (sum(x * x for x in diff)) ** 0.5


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = distance_between_vectors(v1, v2)
print(f"Distance between {v1} & {v2} is {round(result, 2)}")
```

```python
# Zero Vector Check
def is_zero_vector(v):
    return all(x == 0 for x in v)


v = [0, 0, 0]

result = is_zero_vector(v)
print(f"{v} is a zero vector: {result}")
```

```python
# Unit Vector
def unit_vector(v, precision=2):
    norm = vector_norm(v)

    if isinstance(norm, str):
        return norm

    if norm == 0:
        return "Zero vector has no unit vector"

    return [round(x / norm, precision) for x in v]


v = [3, 5, 7]
u = unit_vector(v)
print(f"Unit Vector of {v} is {u}")
```

```python
# Check if Vector is Unit Vector
def is_unit_vector(v):
    return round(vector_norm(v), 6) == 1


v = [0.6, 0.8]

result = is_unit_vector(v)

print(f"{v} is a unit vector: {result}")
```

```python
# Vector Equality
def are_equal(v1, v2):
    return v1 == v2


v1 = [1, 2, 3]
v2 = [1, 2, 3]

print(f"{v1} and {v2} are equal: {are_equal(v1, v2)}")
```

```python
# Parallel Vector Check
def is_parallel(v1, v2):
    if len(v1) != len(v2):
        return "Vectors must have same length"

    ratios = []

    for a, b in zip(v1, v2):
        if b == 0:
            if a != 0:
                return False
        else:
            ratios.append(a / b)

    return len({round(r, 6) for r in ratios}) <= 1


v1 = [1, 2, 3]
v2 = [2, 4, 6]

result = is_parallel(v1, v2)

print(f"{v1} and {v2} are parallel: {result}")
```

```python
# Orthogonal Check
def is_orthogonal(v1, v2):
    dot = vector_dot_product(v1, v2)

    if isinstance(dot, str):
        return dot

    if dot == 0:
        return "Vectors are orthogonal"

    return "Vectors are not orthogonal"


v1 = [2, 5, 8]
v2 = [4, 5, 6]

print(is_orthogonal(v1, v2))
```

```python
# Angle Between Two Vectors and Angle Classification
import math


def angle_between(v1, v2):
    dp = vector_dot_product(v1, v2)

    if isinstance(dp, str):
        return dp

    norm1 = vector_norm(v1)
    norm2 = vector_norm(v2)

    if norm1 == 0 or norm2 == 0:
        return "Zero vector has no defined angle"

    cos_theta = dp / (norm1 * norm2)

    return math.degrees(math.acos(cos_theta))


def classify_angle(v1, v2):
    theta = angle_between(v1, v2)

    if isinstance(theta, str):
        return theta

    if math.isclose(theta, 90, abs_tol=1e-6):
        return f"{theta:.2f}° → Right angle"
    elif theta < 90:
        return f"{theta:.2f}° → Acute angle"
    else:
        return f"{theta:.2f}° → Obtuse angle"


v1 = [2, 5, 8]
v2 = [4, 5, 6]

print(classify_angle(v1, v2))
```

```python
# Cosine Similarity
def cosine_similarity(v1, v2):
    dp = vector_dot_product(v1, v2)

    if isinstance(dp, str):
        return dp

    norm1 = vector_norm(v1)
    norm2 = vector_norm(v2)

    if norm1 == 0 or norm2 == 0:
        return "Zero vector not allowed"

    return round(dp / (norm1 * norm2), 4)


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = cosine_similarity(v1, v2)
print(f"Cosine similarity between {v1} and {v2} is {result}")
```

```python
def verify_schwarz(v1, v2):
    dp = vector_dot_product(v1, v2)

    if isinstance(dp, str):
        return dp

    dp = abs(dp)

    rhs = vector_norm(v1) * vector_norm(v2)

    return dp <= rhs


v1 = [1, 2, 3]
v2 = [4, 5, 6]

result = verify_schwarz(v1, v2)
print(f"Schwarz inequality satisfied: {result}")
```

```python
# Projection of One Vector onto Another
def projection(u, v):
    dp = vector_dot_product(u, v)

    if isinstance(dp, str):
        return dp

    norm_sq = vector_norm(v) ** 2

    if norm_sq == 0:
        return "Cannot project onto zero vector"

    scalar = dp / norm_sq

    return [round(scalar * x, 2) for x in v]


u = [3, 4]
v = [1, 0]

result = projection(u, v)

print(f"Projection of {u} onto {v} is {result}")
```
