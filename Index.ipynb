```python
from sympy import Matrix, latex
from IPython.display import display, Math
import matplotlib.pyplot as plt

import matplotlib as mpl

mpl.rcParams["font.family"] = "Fira Code"
mpl.rcParams["font.size"] = 12
mpl.rcParams["font.weight"] = "bold"

u = Matrix([1, 2])
v = Matrix([3, 4])
k = 5

display(Math("u = " + latex(u) + ", \\; v = " + latex(v)))
display(Math("u + v = " + latex(u + v)))
display(Math("u - v = " + latex(u - v)))
display(Math("u \\cdot v = " + latex(u.dot(v))))
display(Math("k = " + str(k)))
display(Math("k u = " + latex(k * u)))
display(Math("\\tfrac{1}{2} u = " + latex((1 / 2) * u)))

elements = list(u) + list(v)

min_raw = min(elements)
max_raw = max(elements)

if min_raw == 1:
    min_val = -1
else:
    min_val = float(min_raw - 1)

max_val = float(max_raw + 1)

plt.quiver(
    0,
    0,
    float(u[0]),
    float(u[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="r",
    label=r"$\vec{u}$",
)
plt.quiver(
    0,
    0,
    float(v[0]),
    float(v[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="g",
    label=r"$\vec{v}$",
)

plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.xlim(min_val, max_val)
plt.ylim(min_val, max_val)
plt.xlabel("X Axis", fontsize=12, fontweight="bold", family="Fira Code")
plt.ylabel("Y Axis", fontsize=12, fontweight="bold", family="Fira Code")
plt.title(
    r"2D Vectors $\vec{u}$ & $\vec{v}$",
    fontsize=15,
    fontweight="bold",
)
plt.legend(prop={"size": 12, "weight": "bold", "family": "Fira Code"})
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
%matplotlib widget

from sympy import Matrix, latex
from IPython.display import display, Math
import matplotlib.pyplot as plt

u = Matrix([1, 2, 3])
v = Matrix([4, 5, 6])
k = 2

display(Math("u = " + latex(u) + ", \\; v = " + latex(v)))
display(Math("u + v = " + latex(u + v)))
display(Math("u - v = " + latex(u - v)))
display(Math("u \\cdot v = " + latex(u.dot(v))))
display(Math("u \\times v = " + latex(u.cross(v))))
display(Math("k = " + str(k)))
display(Math("k u = " + latex(k * u)))
display(Math("\\tfrac{1}{2} u = " + latex((1 / 2) * u)))

all_vals = list(u) + list(v)
min_raw = min(all_vals)
max_raw = max(all_vals)

min_val = -1 if min_raw == 1 else float(min_raw - 1)
max_val = float(max_raw + 1)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1, projection="3d")
ax.quiver(
    0,
    0,
    0,
    *map(float, u),
    color="red",
    arrow_length_ratio=0.1,
    linewidth=2,
    label=r"$\vec{u}$"
)
ax.quiver(
    0,
    0,
    0,
    *map(float, v),
    color="green",
    arrow_length_ratio=0.05,
    linewidth=2,
    label=r"$\vec{v}$"
)
ax.set_xlim(min_val, max_val)
ax.set_ylim(min_val, max_val)
ax.set_zlim(min_val, max_val)
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_box_aspect([1, 1, 1])
ax.legend()
plt.show()
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

A = Matrix([[2, -1], [1, 2]])

L, U, perm = A.LUdecomposition()
display(
    Math(
        "A = P \\cdot L \\cdot U, \\; P = "
        + latex(perm)
        + ", \\; L = "
        + latex(L)
        + ", \\; U = "
        + latex(U)
    )
)

Q, R = A.QRdecomposition()
display(Math("A = Q \\cdot R, \\; Q = " + latex(Q) + ", \\; R = " + latex(R)))

M = A * A.T
sum = M.cholesky()
display(Math(latex(M) + " = C \\cdot C^T, \\; C = " + latex(sum)))

U, S, V = A.singular_value_decomposition()
display(
    Math(
        "A = U \\cdot \\Sigma \\cdot V^T, \\; U = "
        + latex(U)
        + ", \\; \\Sigma = "
        + latex(S)
        + ", \\; V = "
        + latex(V)
    )
)
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

b1 = Matrix([1, 1])
b2 = Matrix([-1, 1])
P = Matrix.hstack(b1, b2)
coords_new = Matrix([2, 3])
v_standard = P * coords_new

display(Math("P = " + latex(P)))
display(Math("C = " + latex(coords_new)))
display(Math("v = P \\cdot C = " + latex(v_standard)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

A = Matrix([[5, 4], [1, 2]])

P, D = A.diagonalize()

display(Math("A = " + latex(A)))
display(Math("P = " + latex(P)))
display(Math("D = " + latex(D)))
display(Math("A = P D P^{-1}"))
display(Math("P^{-1} A P = D"))
```
