```python
from sympy import Matrix, latex
from IPython.display import display, Math

A = Matrix([[2, 1], [1, 1]])
b = Matrix([3, 2])
x = A.LUsolve(b)

display(
    Math(
        latex(A)
        + "\\cdot x = "
        + latex(b)
        + "\\quad \\Rightarrow \\quad x = "
        + latex(x)
    )
)
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

A = Matrix([[2, -1], [1, 2]])

L, U, perm = A.LUdecomposition()
display(
    Math(
        "A = P \\cdot L \\cdot U, \\; P = "
        + latex(perm)
        + ", \\; L = "
        + latex(L)
        + ", \\; U = "
        + latex(U)
    )
)

Q, R = A.QRdecomposition()
display(Math("A = Q \\cdot R, \\; Q = " + latex(Q) + ", \\; R = " + latex(R)))

M = A * A.T
sum = M.cholesky()
display(Math(latex(M) + " = C \\cdot C^T, \\; C = " + latex(sum)))

U, S, V = A.singular_value_decomposition()
display(
    Math(
        "A = U \\cdot \\Sigma \\cdot V^T, \\; U = "
        + latex(U)
        + ", \\; \\Sigma = "
        + latex(S)
        + ", \\; V = "
        + latex(V)
    )
)
```
