```python
from sympy import Matrix, init_printing
from IPython.display import display

init_printing()

v = Matrix([1, 2, 3, 4])

display(v)

print(f"Vector Shape: {v.shape}")
print(f"Vector Rows: {v.rows}")
print(f"Vector Cols: {v.cols}")
print(f"Total elements: {v.rows * v.cols}\n")
print(f"First element: {v[0]}")
print(f"Last element: {v[3]}")
```

```python
from sympy import Matrix, acos, sqrt, deg

a = Matrix([1, 0])
b = Matrix([1, 1])

angle_in_radian = acos(a.dot(b) / sqrt(a.dot(a) * b.dot(b)))

angle_in_degree = deg(angle_in_radian)

angle_in_radian, angle_in_degree
```

```python
from sympy import Matrix


def projection(v, w):
    return (v.dot(w) / w.dot(w)) * w


def perpendicular(v, w):
    return v - projection(v, w)


v = Matrix([4, 2])
w = Matrix([1, 1])

projection(v, w), perpendicular(v, w)
```

```python
import numpy as np
import matplotlib.pyplot as plt

V = np.array([[1, 2], [2, 1], [3, 4], [-2, 2]])
origin = np.zeros((len(V), 2))

fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-3, 5)
ax.set_ylim(-1, 5)

ax.axhline(0, color="black", alpha=0.25)
ax.axvline(0, color="black", alpha=0.25)
ax.grid(True, linestyle="--", alpha=0.4)

colors = ["red", "blue", "green", "purple"]

ax.quiver(
    origin[:, 0],
    origin[:, 1],
    V[:, 0],
    V[:, 1],
    angles="xy",
    scale_units="xy",
    scale=1,
    color=colors,
)

offsets = [
    (-15, 5),
    (-10, 5),
    (-15, 5),
    (-15, 5),
]

for i, v in enumerate(V):
    ax.annotate(
        f"({v[0]}, {v[1]})",
        (v[0], v[1]),
        textcoords="offset points",
        xytext=offsets[i],
        color=colors[i],
    )

plt.show()
```
