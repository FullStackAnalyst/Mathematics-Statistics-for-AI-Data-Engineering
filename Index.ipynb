```python
from sympy import Matrix, latex
from IPython.display import display, Math

matrix_A = Matrix([[1, 2], [3, 4]])
matrix_B = Matrix([[5, 6], [7, 8]])
Addition = matrix_A + matrix_B

display(Math(latex(matrix_A) + " + " + latex(matrix_B) + " = " + latex(Addition)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

matrix_C = Matrix([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
matrix_D = Matrix([[10, 11, 12], [13, 14, 15], [16, 17, 18]])
Addition = matrix_C + matrix_D

display(Math(latex(matrix_C) + " + " + latex(matrix_D) + " = " + latex(Addition)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

matrix_E = Matrix([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
matrix_F = Matrix([[10, 11, 12], [13, 14, 15], [16, 17, 18]])
Product = matrix_E * matrix_F

display(Math(latex(matrix_E) + " \\times " + latex(matrix_F) + " = " + latex(Product)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

matrix_G = Matrix([[1, 2, 3], [0, 1, 4], [5, 1, 0]])
Inverse = matrix_G.inv()

display(Math(latex(matrix_G) + "^{-1} = " + latex(Inverse)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

matrix_H = Matrix([[1, 2, 9], [4, 5, 6], [7, 8, 3]])
Transpose = matrix_H.T

display(Math(latex(matrix_H) + "^T = " + latex(Transpose)))
```
