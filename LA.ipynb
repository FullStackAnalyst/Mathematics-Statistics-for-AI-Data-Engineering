```python
def get_axis_limits(*vectors, padding=1.0):
    coords_by_dim = list(zip(*vectors))

    limits = []
    for coords in coords_by_dim:
        limits.append(float(min(coords)) - padding)
        limits.append(float(max(coords)) + padding)

    return tuple(limits)
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math
import matplotlib.pyplot as plt

# Define vectors
o = Matrix([0, 0])
u = Matrix([1, 4])
v = Matrix([4, 1])

# Display symbolic representation
display(Math("u = " + latex(u) + ",\\ v = " + latex(v)))

# Use reusable method for axis limits
x_min, x_max, y_min, y_max = get_axis_limits(o, u, v, padding=1.0)

# Plot vectors directly
plt.quiver(
    float(o[0]),
    float(o[1]),
    float(u[0]),
    float(u[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="r",
    label=r"$\vec{u}$",
)

plt.quiver(
    float(o[0]),
    float(o[1]),
    float(v[0]),
    float(v[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="g",
    label=r"$\vec{v}$",
)

# Axes setup
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.title(r"2D Vectors $\vec{u}$ & $\vec{v}$", fontsize=15, fontweight="bold")
plt.tick_params(axis="both", which="major", labelsize=10)

# Legend fix: explicit handles
handles = [
    plt.Line2D([0], [0], color="r", lw=2, label=r"$\vec{u}$"),
    plt.Line2D([0], [0], color="g", lw=2, label=r"$\vec{v}$"),
]
plt.legend(
    handles=handles,
    prop={"size": 12, "weight": "bold"},
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    borderaxespad=0,
    frameon=True,
)

plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math
import matplotlib.pyplot as plt

# Define vectors
o = Matrix([0, 0])
u = Matrix([1, 4])
v = Matrix([4, 1])
neg_v = -v
result = u - v  # subtraction

# Display symbolic representation
display(Math("u = " + latex(u) + ",\\ v = " + latex(v) + ",\\ u-v = " + latex(result)))

# Use reusable method for axis limits
x_min, x_max, y_min, y_max = get_axis_limits(o, u, v, neg_v, result, padding=1.0)

# Plot Subtraction Rule directly
plt.quiver(
    float(o[0]),
    float(o[1]),
    float(u[0]),
    float(u[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="r",
    label=r"$\vec{u}$",
)

plt.quiver(
    float(o[0]),
    float(o[1]),
    float(v[0]),
    float(v[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="g",
    label=r"$\vec{v}$",
)

plt.quiver(
    float(o[0]),
    float(o[1]),
    float(neg_v[0]),
    float(neg_v[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="orange",
    label=r"$-\vec{v}$",
    alpha=0.7,
    linestyle="dashed",
)

plt.quiver(
    float(o[0]),
    float(o[1]),
    float(result[0]),
    float(result[1]),
    angles="xy",
    scale_units="xy",
    scale=1,
    color="b",
    label=r"$\vec{u}-\vec{v}$",
)

# Axes setup
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.title(r"Subtraction Rule of Vectors", fontsize=15, fontweight="bold")
plt.tick_params(axis="both", which="major", labelsize=10)

# Legend fix: explicit handles
handles = [
    plt.Line2D([0], [0], color="r", lw=2, label=r"$\vec{u}$"),
    plt.Line2D([0], [0], color="g", lw=2, label=r"$\vec{v}$"),
    plt.Line2D([0], [0], color="orange", lw=2, label=r"$-\vec{v}$"),
    plt.Line2D([0], [0], color="b", lw=2, label=r"$\vec{u}-\vec{v}$"),
]

plt.legend(
    handles=handles,
    prop={"size": 12, "weight": "bold"},
    loc="upper left",
    bbox_to_anchor=(1.02, 1),
    borderaxespad=0,
    frameon=True,
)

plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
%matplotlib widget

import matplotlib.pyplot as plt
from sympy import Matrix, latex
from IPython.display import display, Math

# Vectors
o = Matrix([0, 0, 0])
u = Matrix([1, 4, 2])
v = Matrix([4, 1, 3])

# Display symbolic math
display(Math("u = " + latex(u) + ",\\ v = " + latex(v)))

# Figure + Axes
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# Plot vectors (loop for clean reuse)
for vec, color, label in [(u, "r", r"$\vec{u}$"), (v, "g", r"$\vec{v}$")]:
    ax.quiver(*o, *vec, color=color, label=label)

# Auto limits
x_min, x_max, y_min, y_max, z_min, z_max = get_axis_limits(o, u, v, padding=1.0)
ax.set(xlim=(x_min, x_max), ylim=(y_min, y_max), zlim=(z_min, z_max))

# Origin axes lines
ax.plot([x_min, x_max], [0, 0], [0, 0], color="black", linewidth=0.5)
ax.plot([0, 0], [y_min, y_max], [0, 0], color="black", linewidth=0.5)
ax.plot([0, 0], [0, 0], [z_min, z_max], color="black", linewidth=0.5)

# Labels + ticks
ax.set_xlabel('X', fontweight="bold")
ax.set_ylabel('Y', fontweight="bold")
ax.set_zlabel('Z', fontweight="bold")
ax.tick_params(labelsize=10)

# Legend (auto from labels)
ax.legend(prop={"size": 12}, loc="upper left", bbox_to_anchor=(1.02, 1), borderaxespad=0, frameon=True)

# Title
ax.set_title(r"3D Vectors $\vec{u},\ \vec{v}$", fontsize=15, fontweight="bold")

plt.show()
```

```python
%matplotlib widget

from sympy import Matrix, latex
from IPython.display import display, Math
import matplotlib.pyplot as plt

u = Matrix([1, 2, 3])
v = Matrix([4, 5, 6])
k = 2

display(Math("u = " + latex(u) + ", \\; v = " + latex(v)))
display(Math("u + v = " + latex(u + v)))
display(Math("u - v = " + latex(u - v)))
display(Math("u \\cdot v = " + latex(u.dot(v))))
display(Math("u \\times v = " + latex(u.cross(v))))
display(Math("k = " + str(k)))
display(Math("k u = " + latex(k * u)))
display(Math("\\tfrac{1}{2} u = " + latex((1 / 2) * u)))

all_vals = list(u) + list(v)
min_raw = min(all_vals)
max_raw = max(all_vals)

min_val = -1 if min_raw == 1 else float(min_raw - 1)
max_val = float(max_raw + 1)

fig = plt.figure()
ax = fig.add_subplot(1, 1, 1, projection="3d")
ax.quiver(
    0,
    0,
    0,
    *map(float, u),
    color="red",
    arrow_length_ratio=0.1,
    linewidth=2,
    label=r"$\vec{u}$"
)
ax.quiver(
    0,
    0,
    0,
    *map(float, v),
    color="green",
    arrow_length_ratio=0.05,
    linewidth=2,
    label=r"$\vec{v}$"
)
ax.set_xlim(min_val, max_val)
ax.set_ylim(min_val, max_val)
ax.set_zlim(min_val, max_val)
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")
ax.set_zlabel("Z Axis")
ax.set_box_aspect([1, 1, 1])
ax.legend()
plt.show()
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

A = Matrix([[2, -1], [1, 2]])

L, U, perm = A.LUdecomposition()
display(
    Math(
        "A = P \\cdot L \\cdot U, \\; P = "
        + latex(perm)
        + ", \\; L = "
        + latex(L)
        + ", \\; U = "
        + latex(U)
    )
)

Q, R = A.QRdecomposition()
display(Math("A = Q \\cdot R, \\; Q = " + latex(Q) + ", \\; R = " + latex(R)))

M = A * A.T
sum = M.cholesky()
display(Math(latex(M) + " = C \\cdot C^T, \\; C = " + latex(sum)))

U, S, V = A.singular_value_decomposition()
display(
    Math(
        "A = U \\cdot \\Sigma \\cdot V^T, \\; U = "
        + latex(U)
        + ", \\; \\Sigma = "
        + latex(S)
        + ", \\; V = "
        + latex(V)
    )
)
```
